```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# ============================
# (1) Gerar dados sintéticos
# ============================
np.random.seed(42)
N = 300  # número total de pontos
centros = [(2, 2), (8, 3), (5, 8)]  # centros "reais" dos clusters

X, _ = make_blobs(n_samples=N, centers=centros, cluster_std=1.0, random_state=42)

# ============================
# (2) Aplicar KMeans
# ============================
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X)
centroids = kmeans.cluster_centers_

# ============================
# (3) Criar gráfico
# ============================
fig, ax = plt.subplots(figsize=(6, 6))

# Paleta de cores simples
colors = ["tab:blue", "tab:orange", "tab:green"]

# Plotar cada cluster
for i, color in enumerate(colors):
    ax.scatter(
        X[labels == i, 0],
        X[labels == i, 1],
        s=40,
        c=color,
        label=f"Cluster {i+1}",
        alpha=0.7,
        edgecolor="k",
        linewidth=0.5
    )

# Plotar os centróides
ax.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker="X",
    s=150,
    c="red",
    label="Centróides",
    edgecolor="black"
)

# ============================
# (4) Configurações de estilo
# ============================
ax.legend(loc="upper left", frameon=True, fontsize=9)
ax.set_xlabel("Componente 1")
ax.set_ylabel("Componente 2")

ax.tick_params(direction='in', length=4, width=1.0, top=True, bottom=True, left=True, right=True)
ax.xaxis.set_major_locator(MaxNLocator(integer=False))
ax.yaxis.set_major_locator(MaxNLocator(integer=False))

plt.tight_layout()
pathfg = "clusterizacao_3clusters.png"
plt.savefig(pathfg, dpi=300, bbox_inches="tight")
print(f"Gráfico salvo em: {pathfg}")
plt.show()
```
