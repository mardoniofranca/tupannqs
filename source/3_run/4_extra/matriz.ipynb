```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import MaxNLocator

# ======================
# 1. Criar matriz de dados (10x10) como no exemplo anterior
# ======================
data = np.array([
    [ 1,  2,  1,  5,  3,  7, 10, -9,  4,  8],
    [ 2,  4,  2,  6,  5,  6,  9, -8,  7,  5],
    [ 3,  6,  2,  7,  6,  5, 12, -7,  2,  9],
    [ 4,  8,  3,  8,  8,  7, 11, -6,  5,  6],
    [ 5, 10,  4,  9,  7,  6, 13, -5,  8,  7],
    [ 6, 12,  5, 10,  6,  8, 12, -4,  3,  4],
    [ 7, 14,  5, 11,  9,  7, 14, -3,  6,  5],
    [ 8, 16,  6, 12, 10,  9, 15, -2,  5,  6],
    [ 9, 18,  6, 13,  8, 10, 13, -1,  7,  5],
    [10, 20,  7, 14, 11,  9, 16,  0,  4,  8]
])

# Converter em DataFrame para facilitar
cols = [f"C{i}" for i in range(1, 11)]
df = pd.DataFrame(data, columns=cols)

# ======================
# 2. Calcular matriz de correlação de Pearson
# ======================
corr_matrix = df.corr()

# ======================
# 3. Plotar graficamente no estilo solicitado
# ======================
fig, axs = plt.subplots(3, 1, figsize=(6, 10))

# ===== (a) Heatmap completo =====
ax = axs[0]
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True, ax=ax)
ax.set_title("(a) Matriz de correlação completa", fontsize=10)
ax.tick_params(direction='in', length=4, width=1.0, top=False, bottom=False, left=False, right=False)

# ===== (b) Subconjunto com correlação positiva (C1–C4) =====
ax = axs[1]
sns.heatmap(corr_matrix.loc["C1":"C6", "C1":"C6"], annot=True, fmt=".2f", cmap="Reds", cbar=True, ax=ax)
ax.set_title("(b) Submatriz: correlações positivas", fontsize=10)
ax.tick_params(direction='in', length=4, width=1.0, top=False, bottom=False, left=False, right=False)

# ===== (c) Subconjunto com correlação negativa (C7–C8) =====
ax = axs[2]
sns.heatmap(corr_matrix.loc["C8":"C10", "C8":"C10"], annot=True, fmt=".2f", cmap="Blues", cbar=True, ax=ax)
ax.set_title("(c) Submatriz: correlações negativas", fontsize=10)
ax.tick_params(direction='in', length=4, width=1.0, top=False, bottom=False, left=False, right=False)

# Ajustes finais
plt.tight_layout()


pathfg = "pearson_mateix.png"
print(pathfg)
plt.savefig(pathfg, dpi=300, bbox_inches='tight')


plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ======================
# 1. Criar a tabela base 10x10
# ======================
data = np.array([
    [ 1,  2,  1,   5,   3,  7, 10,   1,  -1.2,   0], 
    [ 2,  4,  2,   6,   5,  6,  9,   2,  -2.1,   0],
    [ 3,  6,  2,   7,   6,  5, 12,   3,  -2.9,   0],
    [ 4,  8,  3,   8,   8,  7, 11,   4,  -4.2,   0],
    [ 5, 10,  4,   9,   7,  6, 13,   5,  -5.1,   0],
    [ 6, 12,  5,  10,   6,  8, 12,   6,  -6.0,   0],
    [ 7, 14,  5,  11,   9,  7, 14,   7,  -7.2,   0],
    [ 8, 16,  6,  12,  10,  9, 15,   8,  -8.1,   0],
    [ 9, 18,  6,  13,   8,  6, 13,   9,  -9.0,   0],
    [10, 18,  7, 12.5, 11,  9, 14,   9,  -9.8,   0],
])

# ======================
# 2. Converter em DataFrame
# ======================
cols = [f"C{i}" for i in range(1, 11)]
df = pd.DataFrame(data, columns=cols)

# ======================
# 3. Corrigir C10 para ser ~ -0.9 correlacionada com C9
# ======================
np.random.seed(42)
df["C10"] = -1 * df["C9"] + np.random.normal(0, 0.5, size=len(df))

# ======================
# 4. Matriz de correlação
# ======================
corr_matrix = df.corr()

print("Matriz de Correlação de Pearson:")
print(corr_matrix.round(2))

# ======================
# 5. Plotar heatmap
# ======================
plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0, cbar=True,
            linewidths=0.5, linecolor='gray')

plt.title("Matriz de Correlação de Pearson (10x10)", fontsize=12)
plt.tight_layout()
plt.show()
```
