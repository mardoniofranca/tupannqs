```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

# Gerar dados aleatórios
np.random.seed(42)
N = 100

# Correlação positiva
x_pos = np.linspace(0, 10, N)
y_pos = 2 * x_pos + np.random.normal(0, 3, N)

# Correlação negativa
x_neg = np.linspace(0, 10, N)
y_neg = -2 * x_neg + np.random.normal(0, 3, N)

# Correlação neutra
x_neu = np.random.uniform(0, 10, N)
y_neu = np.random.uniform(0, 10, N)

# Criar subplots
fig, axs = plt.subplots(3, 1, figsize=(6, 8), sharex=True)

# ===== (a) Correlação positiva =====
ax = axs[0]
ax.scatter(x_pos, y_pos, color="tab:blue", label="(a) Correlação positiva", alpha=0.7)
ax.legend(loc="upper right", frameon=True, fontsize=9)
ax.set_ylabel("y")
ax.tick_params(direction='in', length=4, width=1.0, top=True, bottom=True, left=True, right=True)
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
#ax.text(0.05, 0.9, "(a) Correlação positiva", transform=ax.transAxes, fontsize=10, verticalalignment="top")

# ===== (b) Correlação negativa =====
ax = axs[1]
ax.scatter(x_neg, y_neg, color="tab:red", label="(b) Correlação negativa", alpha=0.7)
ax.legend(loc="upper left", frameon=True, fontsize=9)
ax.set_ylabel("y")
ax.tick_params(direction='in', length=4, width=1.0, top=True, bottom=True, left=True, right=True)
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
#ax.text(0.05, 0.9, "(b) Correlação negativa", transform=ax.transAxes, fontsize=10, verticalalignment="top")

# ===== (c) Correlação neutra =====
ax = axs[2]
ax.scatter(x_neu, y_neu, color="tab:green", label="(c) Correlação neutra", alpha=0.7)
ax.legend(loc="upper left", frameon=True, fontsize=9)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.tick_params(direction='in', length=4, width=1.0, top=True, bottom=True, left=True, right=True)
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
#ax.text(0.05, 0.9, "(c) Correlação neutra", transform=ax.transAxes, fontsize=10, verticalalignment="top")

# Ajustes finais
plt.subplots_adjust(hspace=0.3)
pathfg = "pearson.png"
print(pathfg)
plt.savefig(pathfg, dpi=300, bbox_inches='tight')
plt.show()
```
