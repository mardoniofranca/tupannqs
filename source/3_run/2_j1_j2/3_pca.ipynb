### PCA

```python
L = 8; wT = 210; it = 500; ll = str(L)
kernel_bias = 'kernel'; real_img = 'real'; stride = 5
path    = fr"data/server/L_{L}/it_{it}_l_{L}"; ex_0    = "00"
extEgsL = []; itsNetL = []; netEgsL = []
dfs     = []

for j, wT in enumerate(wTs):
    
    ex_n = iLs[j];    
    df_w = w_net(path,ex_n,L,wT,it,kernel_bias,real_img);

    exact_gs_energy             = f_exact_gs_energy(path,ex_0,ll,wT)
    iters_Net_ds, energy_Net_ds = f_ffnn(path,ex_n, ll, wT, it,stride)


    dfs.append(df_w)

    extEgsL.append(exact_gs_energy)
    itsNetL.append(iters_Net_ds)
    energy_Net_ds.append(energy_Net_ds)   
```

```python
show = 1; path_img = f"fig/L_{L}/PCA_2D/"; T = "3_R"
plot_n_2d_pca(dfs[0], dfs[1], dfs[2],
              wTs,iLs,
              0, -1, 
              ll,it,"FFNN",T,
              1,path_img)
```
