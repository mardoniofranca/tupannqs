```python
def correlation_matrix_with_pvalues(df):
    cols = df.columns
    n = len(cols)
    corr_matrix = np.zeros((n, n))
    p_matrix = np.zeros((n, n))
    
    for i in range(n):
        for j in range(i, n):  # Apenas metade superior para eficiência
            if i == j:
                corr_matrix[i, j] = 1.0
                p_matrix[i, j] = 0.0
            else:
                corr, p_val = pearsonr(df.iloc[:, i].dropna(), df.iloc[:, j].dropna())
                corr_matrix[i, j] = corr
                corr_matrix[j, i] = corr  
                p_matrix[i, j] = p_val
                p_matrix[j, i] = p_val   #
    
    return pd.DataFrame(corr_matrix, index=cols, columns=cols), pd.DataFrame(p_matrix, index=cols, columns=cols)
```

```python
def corr_metrics(df):
    corr_matrix,p_matrix   = correlation_matrix_with_pvalues(df)
    soma_correlacoes       = corr_matrix.values.sum()
    corr_values            = corr_matrix.values    
    
    np.fill_diagonal(corr_values, np.nan)     
    
    media_correlacoes     = np.nanmean(corr_values)
    dispersao_correlacoes = np.nanstd(corr_values)
    p_values              = p_matrix.values
    media_pvalores        = np.nanmean(p_values)
    
    return soma_correlacoes, dispersao_correlacoes, media_correlacoes, media_pvalores,  np.sum(~np.isnan(corr_values)),   np.nanmax(corr_values),np.nanmin(corr_values),np.mean(p_values < 0.05)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_circle_angles(angles_deg):
    # Conversão para radianos
    angles_rad = np.radians(angles_deg)

    # Coordenadas dos pontos
    x = np.cos(angles_rad)
    y = np.sin(angles_rad)

    # Criar figura
    fig, ax = plt.subplots(figsize=(6,6))
    ax.set_aspect('equal')

    # Círculo unitário
    circle = plt.Circle((0,0), 1, color='lightgray', fill=False, linestyle='--')
    ax.add_artist(circle)

    # Eixos
    ax.axhline(0, color='black', linewidth=0.8)
    ax.axvline(0, color='black', linewidth=0.8)

    # Plotar vetores
    for angle, xi, yi in zip(angles_deg, x, y):
        ax.arrow(0, 0, xi, yi, head_width=0.05, head_length=0.07,
                 fc='blue', ec='blue', alpha=0.8)
        ax.text(xi*1.1, yi*1.1, f"{angle}°", fontsize=10, ha='center')

    # Limites
    ax.set_xlim(-1.2, 1.2)
    ax.set_ylim(-1.2, 1.2)

    # Configurações
    ax.set_title("Círculo Trigonométrico com Ângulos")
    ax.set_xticks([])
    ax.set_yticks([])

    plt.show()

# Exemplo
plot_circle_angles([15, 60, 210])
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import json
import numpy as np
from matplotlib.ticker import MaxNLocator, FixedLocator

def plot_energy_convergence(exact_energy_paths, calculated_energy_files, 
                           thetas, L, IT, network_labels, 
                           x_pos=0.4, y_pos=0.94, str_nets=''):
    """
    Plota a convergência de energia para múltiplos ângulos theta e métodos
    
    Parameters:
    -----------
    exact_energy_paths : list
        Lista de caminhos para arquivos CSV com energias exatas
    calculated_energy_files : list of lists
        Lista de listas com caminhos para arquivos JSON de energias calculadas
        Formato: [[jastrow1, rbm1, ffnn1, nffnn1], [jastrow2, rbm2, ffnn2, nffnn2], ...]
    thetas : list
        Lista de ângulos theta (strings ou valores)
    L : int
        Tamanho do sistema
    IT : int
        Número de iterações
    network_labels : list
        Rótulos para os diferentes métodos [jastrow_label, rbm_label, ffnn_label, nffnn_label]
    x_pos, y_pos : float
        Posição do texto no subplot
    str_nets : str
        String identificadora para o nome do arquivo
    """
    
    n_plots = len(exact_energy_paths)
    
    # Configuração da figura
    fig, axs = plt.subplots(n_plots, 1, figsize=(8, 3 * n_plots), sharex=True)
    
    if n_plots == 1:
        axs = [axs]  # Garante que axs seja sempre uma lista
    
    # Função auxiliar para carregar dados de energia
    def load_energy_data(file_path):
        """Carrega dados de energia de arquivos JSON"""
        try:
            with open(file_path) as f:
                data = json.load(f)
            
            iters = data["Energy"]["iters"]
            energy_raw = data["Energy"]["Mean"]
            
            if isinstance(energy_raw, list):
                energy = [e["real"] if isinstance(e, dict) else e for e in energy_raw]
            elif isinstance(energy_raw, dict):
                energy = energy_raw.get("real", energy_raw)
            else:
                energy = energy_raw
                
            return iters, energy
        except (FileNotFoundError, KeyError, json.JSONDecodeError) as e:
            print(f"Erro ao carregar {file_path}: {e}")
            return [], []
    
    # Função para downsampling
    def downsample(x, y, stride):
        """Reduz a quantidade de pontos para plotagem"""
        if len(x) == 0 or len(y) == 0:
            return [], []
        return x[::stride], y[::stride]
    
    # Estilos de plotagem (defina esses globalmente ou passe como parâmetro)
    linestyles = ['-', '--', '-.', ':']
    markers = ['o', 's', '^', 'D']
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']
    
    # Plot para cada ângulo theta
    for i, (ax, exact_path, calc_files, theta) in enumerate(zip(axs, exact_energy_paths, calculated_energy_files, thetas)):
        
        # Carrega energia exata
        try:
            df = pd.read_csv(exact_path)
            exact_gs_energy = df.iloc[0, 2]
        except (FileNotFoundError, IndexError) as e:
            print(f"Erro ao carregar energia exata {exact_path}: {e}")
            exact_gs_energy = 0
        
        # Processa cada método
        methods_data = []
        for j, file_path in enumerate(calc_files):
            if file_path:  # Verifica se o arquivo foi fornecido
                full_path = file_path if file_path.endswith('.log') else file_path + '.log'
                iters, energy = load_energy_data(full_path)
                if iters and energy:
                    iters_ds, energy_ds = downsample(iters, energy, 5)
                    methods_data.append((iters_ds, energy_ds, network_labels[j], j))
        
        # Plota os dados
        for iters_ds, energy_ds, label, style_idx in methods_data:
            ax.plot(iters_ds, energy_ds, label=label, 
                    linestyle=linestyles[style_idx % len(linestyles)],
                    marker=markers[style_idx % len(markers)], 
                    color=colors[style_idx % len(colors)], 
                    markersize=5, markevery=10)
        
        # Linha da energia exata
        if exact_gs_energy != 0:
            ax.axhline(y=exact_gs_energy, color='k', lw=2, ls='--', label='Exact')
        
        # Configurações do eixo
        ax.tick_params(direction='in', length=4, width=1.0, 
                      top=True, bottom=True, left=True, right=True)
        ax.tick_params(which='minor', direction='in', 
                      length=2, width=0.8, 
                      top=True, bottom=True, left=True, right=True)
        
        ax.xaxis.set_major_locator(MaxNLocator(integer=True))
        ax.yaxis.set_major_locator(MaxNLocator(nbins=5))
        ax.set_ylabel(r"Energia")
        
        # Texto do subplot
        subplot_label = chr(97 + i)  # 'a', 'b', 'c', ...
        ax.text(x_pos, y_pos, f'({subplot_label}) $L={L}$, $\\theta = {theta}$', 
                transform=ax.transAxes, fontsize=12, verticalalignment='top')
        
        # Legenda apenas no primeiro subplot
        if i == 0:
            ax.legend(fontsize=9, loc='upper right', frameon=False)
    
    # Configuração final
    axs[-1].set_xlabel(r"Iterações")
    plt.subplots_adjust(hspace=0.1, wspace=0.1)
    
    # Nome do arquivo de saída
    theta_str = "_".join([str(theta).replace("/", "").replace("\\", "") for theta in thetas])
    pathfg = f"fig/wff/EGS_L_{L}_IT_{IT}_theta_{theta_str}_{str_nets}.png"
    print(f"Salvando: {pathfg}")
    
    plt.savefig(pathfg, dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# Exemplo de uso genérico
if __name__ == "__main__":
    # Defina esses estilos globalmente ou importe de outro arquivo
    linestyles = ['-', '--', '-.', ':']
    markers = ['o', 's', '^', 'D']
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']
    
    # Dados de exemplo - totalmente configurável
    exact_energy_paths = [
        "exact_energy_theta_pi.csv",
        "exact_energy_theta_7pi6.csv",
        "exact_energy_theta_pi2.csv"  # Adicione mais se necessário
    ]
    
    calculated_energy_files = [
        # Para theta = π
        [
            "jastrow_pi",      # Jastrow
            "rbm_pi",          # RBM
            "ffnn_pi",         # FFNN
            "nffnn_pi"         # NFFNN
        ],
        # Para theta = 7π/6
        [
            "jastrow_7pi6",
            "rbm_7pi6", 
            "ffnn_7pi6",
            "nffnn_7pi6"
        ],
        # Para theta = π/2 (exemplo adicional)
        [
            "jastrow_pi2",
            "rbm_pi2",
            "ffnn_pi2",
            "nffnn_pi2"
        ]
    ]
    
    thetas = ["π", "7π/6", "π/2"]
    network_labels = ["JASTROW", "RBM", "FFNN", "NFFNN"]
    
    # Chama a função
    plot_energy_convergence(
        exact_energy_paths=exact_energy_paths,
        calculated_energy_files=calculated_energy_files,
        thetas=thetas,
        L=10,
        IT=1000,
        network_labels=network_labels,
        x_pos=0.4,
        y_pos=0.94,
        str_nets="comparison"
    )
```
