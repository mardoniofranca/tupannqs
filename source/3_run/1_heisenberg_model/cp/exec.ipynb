```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_pearson_neurons(paths1, paths2, point_idx, save_path="pearson_neurons.png"):
    """
    paths1: lista de arquivos CSV do regime ferromagnético
    paths2: lista de arquivos CSV do regime antiferromagnético
    point_idx: índice da linha (tempo/iterações) que deseja analisar
    """
    # --- Função auxiliar ---
    def prepare_corr_matrix(path, idx):
        df = pd.read_csv(path)
        data = df.iloc[idx, 1:].astype(float)  # remove 'id'
        df_neurons = pd.DataFrame([data.values], columns=[f"N{i+1}" for i in range(len(data))])
        return df_neurons.T.corr()  # correlação entre todos os neurônios

    # Calcular correlações para cada regime
    corr1 = pd.read_csv(paths1[2]).iloc[:, 1:].corr(method="pearson")
    corr2 = pd.read_csv(paths2[2]).iloc[:, 1:].corr(method="pearson")

    # --- Plot ---
    fig, axs = plt.subplots(1, 2, figsize=(14, 6))

    sns.heatmap(corr1, cmap=sns.diverging_palette(220, 10, as_cmap=True),
                vmin=-1, vmax=1, square=True, annot=False, ax=axs[0])
    axs[0].set_title("Ferromagnetismo - Correlação entre neurônios")

    sns.heatmap(corr2, cmap=sns.diverging_palette(220, 10, as_cmap=True),
                vmin=-1, vmax=1, square=True, annot=False, ax=axs[1])
    axs[1].set_title("Antiferromagnetismo - Correlação entre neurônios")

    plt.tight_layout()
    plt.savefig(save_path, dpi=300)
    plt.show()

    print(f"Figura salva em: {save_path}")
    return corr1, corr2
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_pearson_side_by_side(paths1, paths2, initial_point_idx, last_point_idx, save_path="pearson_side_by_side.png"):
    # --- Função auxiliar ---
    def prepare_corr_matrix(path, initial_idx, final_idx, prefix):
        df = pd.read_csv(path)
        initial = df.iloc[initial_idx, 1:].astype(float)  # Remove 'id'
        final   = df.iloc[final_idx, 1:].astype(float)
        combined = pd.DataFrame({
            f"{prefix}_Inicial": initial.values,
            f"{prefix}_Final": final.values
        })
        return combined.corr()

    # Calcular correlações para cada regime
    corr1 = prepare_corr_matrix(paths1[2], initial_point_idx, last_point_idx, "Ferro")
    corr2 = prepare_corr_matrix(paths2[2], initial_point_idx, last_point_idx, "Antiferro")

    # --- Plot ---
    fig, axs = plt.subplots(1, 2, figsize=(12, 5))

    sns.heatmap(corr1, cmap=sns.diverging_palette(220, 10, as_cmap=True),
                vmin=-1, vmax=1, square=True, annot=True, fmt=".2f", ax=axs[0])
    axs[0].set_title("Ferromagnetismo (Inicial × Final)")

    sns.heatmap(corr2, cmap=sns.diverging_palette(220, 10, as_cmap=True),
                vmin=-1, vmax=1, square=True, annot=True, fmt=".2f", ax=axs[1])
    axs[1].set_title("Antiferromagnetismo (Inicial × Final)")

    plt.tight_layout()
    plt.savefig(save_path, dpi=300)
    plt.show()

    print(f"Figura salva em: {save_path}")
    return corr1, corr2
```
