```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Função sigmoide
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
###Dados
x = np.linspace(-10, 10, 400)
y = sigmoid(x)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, color='red', linestyle='-', marker='o', markevery=40, label=r'$\sigma(x)$')

# Estilo inspirado nos gráficos fornecidos
plt.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.7)
plt.xlabel(r'$x$', fontsize=12)
plt.ylabel(r'$\sigma(x)$', fontsize=12)
plt.legend(loc='upper left')
plt.tight_layout()

plt.savefig("fig/sigmoide.png", dpi=300, bbox_inches='tight')
```
