```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, FixedLocator

# Dados simulados
h0 = np.linspace(0, 2, 15)
theta_values = [0, 2*np.pi/3, 4*np.pi/3, 5*np.pi/3, np.pi, np.pi/3]
colors = ['blue', 'green', 'magenta', 'black', 'red', 'goldenrod']
markers = ['x', 's', 'D', 'v', '*', 'h']
linestyles = ['-', '--', '-.', ':', '-.', '--']

def simulate_L_bar(h0, theta):
    return 1 / (1 + np.exp(-5*(h0 - 0.8))) + 0.05 * np.sin(theta)

def simulate_S_bar(h0, theta):
    return 2.5 * np.exp(-2 * h0) + 0.2 * np.cos(theta)

# Configuração da figura
fig, axs = plt.subplots(2, 2, figsize=(10, 7), sharex=True)

# ========== Subplot (a) ==========
ax = axs[0, 0]
for i, theta in enumerate(theta_values):
    y = simulate_L_bar(h0, theta) + 0.05 * np.random.rand(len(h0))
    
    if theta == 0:
        label = r"$\theta=0$"
    elif np.isclose(theta, np.pi):
        label = r"$\theta=\pi$"
    elif np.isclose(theta, np.pi/3):
        label = r"$\theta=\pi/3$"
    elif np.isclose(theta, 2*np.pi/3):
        label = r"$\theta=2\pi/3$"
    elif np.isclose(theta, 4*np.pi/3):
        label = r"$\theta=4\pi/3$"
    elif np.isclose(theta, 5*np.pi/3):
        label = r"$\theta=5\pi/3$"
    else:
        label = fr"$\theta={theta:.2f}$"

    ax.plot(h0, y, label=label, linestyle=linestyles[i],
            marker=markers[i], color=colors[i], markersize=5)

# Configurações do eixo
ax.tick_params(direction='in', length=4, width=1.0, top=True, bottom=True, left=True, right=True)
ax.tick_params(which='minor', direction='in', length=2, width=0.8, top=True, bottom=True, left=True, right=True)
ax.tick_params(labelbottom=False)
ax.xaxis.set_major_locator(MultipleLocator(0.5))
ax.xaxis.set_minor_locator(MultipleLocator(0.1))
ax.yaxis.set_major_locator(FixedLocator([0.0, 0.4, 1.0]))
ax.yaxis.set_minor_locator(FixedLocator([]))
ax.set_ylabel(r"$\bar{L}$")
ax.text(0.03, 0.94, '(a) $L=12$', transform=ax.transAxes, fontsize=12, verticalalignment='top')

# ========== Subplot (b) ==========
ax = axs[0, 1]
for i, theta in enumerate(theta_values):
    y = simulate_L_bar(h0, theta) + 0.05 * np.random.rand(len(h0))
    
    if theta == 0:
        label = r"$\theta=0$"
    elif np.isclose(theta, np.pi):
        label = r"$\theta=\pi$"
    elif np.isclose(theta, np.pi/3):
        label = r"$\theta=\pi/3$"
    elif np.isclose(theta, 2*np.pi/3):
        label = r"$\theta=2\pi/3$"
    elif np.isclose(theta, 4*np.pi/3):
        label = r"$\theta=4\pi/3$"
    elif np.isclose(theta, 5*np.pi/3):
        label = r"$\theta=5\pi/3$"
    else:
        label = fr"$\theta={theta:.2f}$"

    ax.plot(h0, y, label=label, linestyle=linestyles[i],
            marker=markers[i], color=colors[i], markersize=5)

# Configurações do eixo
ax.tick_params(direction='in', length=4, width=1.0, top=True, bottom=True, left=True, right=True)
ax.tick_params(which='minor', direction='in', length=2, width=0.8, top=True, bottom=True, left=True, right=True)
ax.tick_params(labelleft=False, labelbottom=False)
ax.xaxis.set_major_locator(MultipleLocator(0.5))
ax.xaxis.set_minor_locator(MultipleLocator(0.1))
ax.yaxis.set_major_locator(MultipleLocator(1.0))
ax.yaxis.set_minor_locator(MultipleLocator(0.2))
ax.text(0.03, 0.94, '(b) $L=20$', transform=ax.transAxes, fontsize=12, verticalalignment='top')
ax.legend(fontsize=9, loc='lower right', frameon=False)

# ========== Subplot (c) ==========
ax = axs[1, 0]
for i, theta in enumerate(theta_values):
    y = simulate_S_bar(h0, theta) + 0.1 * np.random.rand(len(h0))
    
    if theta == 0:
        label = r"$\theta=0$"
    elif np.isclose(theta, np.pi):
        label = r"$\theta=\pi$"
    elif np.isclose(theta, np.pi/3):
        label = r"$\theta=\pi/3$"
    elif np.isclose(theta, 2*np.pi/3):
        label = r"$\theta=2\pi/3$"
    elif np.isclose(theta, 4*np.pi/3):
        label = r"$\theta=4\pi/3$"
    elif np.isclose(theta, 5*np.pi/3):
        label = r"$\theta=5\pi/3$"
    else:
        label = fr"$\theta={theta:.2f}$"

    ax.plot(h0, y, label=label, linestyle=linestyles[i],
            marker=markers[i], color=colors[i], markersize=5)

# Configurações do eixo
ax.tick_params(direction='in', length=4, width=1.0, top=True, bottom=True, left=True, right=True)
ax.tick_params(which='minor', direction='in', length=2, width=0.8, top=True, bottom=True, left=True, right=True)
ax.xaxis.set_major_locator(MultipleLocator(0.5))
ax.xaxis.set_minor_locator(MultipleLocator(0.1))
ax.yaxis.set_major_locator(MultipleLocator(1.0))
ax.yaxis.set_minor_locator(MultipleLocator(0.2))
ax.set_ylabel(r"$\bar{S}_L$")
ax.set_xlabel(r"$h_0$")
ax.text(0.03, 0.94, '(c) $L=12$', transform=ax.transAxes, fontsize=12, verticalalignment='top')

# ========== Subplot (d) ==========
ax = axs[1, 1]
for i, theta in enumerate(theta_values):
    y = simulate_S_bar(h0, theta) + 0.1 * np.random.rand(len(h0))
    
    if theta == 0:
        label = r"$\theta=0$"
    elif np.isclose(theta, np.pi):
        label = r"$\theta=\pi$"
    elif np.isclose(theta, np.pi/3):
        label = r"$\theta=\pi/3$"
    elif np.isclose(theta, 2*np.pi/3):
        label = r"$\theta=2\pi/3$"
    elif np.isclose(theta, 4*np.pi/3):
        label = r"$\theta=4\pi/3$"
    elif np.isclose(theta, 5*np.pi/3):
        label = r"$\theta=5\pi/3$"
    else:
        label = fr"$\theta={theta:.2f}$"

    ax.plot(h0, y, label=label, linestyle=linestyles[i],
            marker=markers[i], color=colors[i], markersize=5)

# Configurações do eixo
ax.tick_params(direction='in', length=4, width=1.0, top=True, bottom=True, left=True, right=True)
ax.tick_params(which='minor', direction='in', length=2, width=0.8, top=True, bottom=True, left=True, right=True)
ax.tick_params(labelleft=False)
ax.xaxis.set_major_locator(MultipleLocator(0.5))
ax.xaxis.set_minor_locator(MultipleLocator(0.1))
ax.yaxis.set_major_locator(MultipleLocator(1.0))
ax.yaxis.set_minor_locator(MultipleLocator(0.2))
ax.set_xlabel(r"$h_0$")
ax.text(0.03, 0.94, '(d) $L=20$', transform=ax.transAxes, fontsize=12, verticalalignment='top')

# Layout final
plt.subplots_adjust(hspace=0.1, wspace=0.1)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, FixedLocator
```

```python
# Dados simulados
h0 = np.linspace(0, 2, 15)
theta_values = [0, 2*np.pi/3, 4*np.pi/3, 5*np.pi/3, np.pi, np.pi/3]
colors = ['blue', 'green', 'magenta', 'black', 'red', 'goldenrod']
markers = ['x', 's', 'D', 'v', '*', 'h']
linestyles = ['-', '--', '-.', ':', '-.', '--']
```

```python
def simulate_L_bar(h0, theta):
    return 1 / (1 + np.exp(-5*(h0 - 0.8))) + 0.05 * np.sin(theta)

def simulate_S_bar(h0, theta):
    return 2.5 * np.exp(-2 * h0) + 0.2 * np.cos(theta)
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 7), sharex=True)

for idx, ax in enumerate(axs.flatten()):
    for i, theta in enumerate(theta_values):
        if idx < 2:
            y = simulate_L_bar(h0, theta) + 0.05 * np.random.rand(len(h0))
        else:
            y = simulate_S_bar(h0, theta) + 0.1 * np.random.rand(len(h0))

        if theta == 0:
            label = r"$\theta=0$"
        elif np.isclose(theta, np.pi):
            label = r"$\theta=\pi$"
        elif np.isclose(theta, np.pi/3):
            label = r"$\theta=\pi/3$"
        elif np.isclose(theta, 2*np.pi/3):
            label = r"$\theta=2\pi/3$"
        elif np.isclose(theta, 4*np.pi/3):
            label = r"$\theta=4\pi/3$"
        elif np.isclose(theta, 5*np.pi/3):
            label = r"$\theta=5\pi/3$"
        else:
            label = fr"$\theta={theta:.2f}$"

        ax.plot(h0, y, label=label, linestyle=linestyles[i],
                marker=markers[i], color=colors[i], markersize=5)

    # Ticks simétricos (todos para dentro)
    ax.tick_params(
        direction='in',
        length=4,
        width=1.0,
        top=True, bottom=True,
        left=True, right=True
    )
    ax.tick_params(which='minor', direction='in', length=2, width=0.8,
                   top=True, bottom=True, left=True, right=True)

    # Remover rótulo dos y da direita e x do topo
    if idx % 2 == 1:
        ax.tick_params(labelleft=False)
    if idx < 2:
        ax.tick_params(labelbottom=False)

    # Eixo X (todos)
    ax.xaxis.set_major_locator(MultipleLocator(0.5))
    ax.xaxis.set_minor_locator(MultipleLocator(0.1))

    # Eixo Y (padrão para todos)
    ax.yaxis.set_major_locator(MultipleLocator(1.0))
    ax.yaxis.set_minor_locator(MultipleLocator(0.2))

# ✅ Subplot (a): Y com apenas 0.0, 0.4, 1.0
axs[0, 0].yaxis.set_major_locator(FixedLocator([0.0, 0.4, 1.0]))
axs[0, 0].yaxis.set_minor_locator(FixedLocator([]))  # Sem minor ticks
axs[0, 0].set_ylabel(r"$\bar{L}$")

# Subplot (c): continua com marcações amplas
axs[1, 0].set_ylabel(r"$\bar{S}_L$")
for ax in axs[1]:
    ax.set_xlabel(r"$h_0$")

# Letras (a)-(d)
annotations = ['(a) $L=12$', '(b) $L=20$', '(c) $L=12$', '(d) $L=20$']
for ax, label in zip(axs.flatten(), annotations):
    ax.text(0.03, 0.94, label, transform=ax.transAxes,
            fontsize=12, verticalalignment='top')

# Legenda apenas no subplot (b)
axs[0, 1].legend(fontsize=9, loc='lower right', frameon=False)

# Layout final
plt.subplots_adjust(hspace=0.1, wspace=0.1)
plt.show()
```
