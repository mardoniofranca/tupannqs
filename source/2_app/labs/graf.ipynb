```python
# Parâmetros
marker_size = 6
markers = ["o", "s", "D", "v", "^", "p", "*"]  # exemplo

# Geração dos theta_values
theta_values = np.linspace(0, 2 * np.pi, 2000)
t_0theta_values = theta_values[::3]

# Geração dos dados de correlação (exemplos usando funções trigonométricas)
correlation_values = np.cos(theta_values)  # t0 = 0.3
t_0_0correlation_values = np.cos(theta_values + 0.3)  # t0 = 0

# Subamostragem
redu_t_0_0correlation_values = t_0_0correlation_values[::3]

# Outra variável de correlação (exemplo com seno ao quadrado)
correlation = np.sin(theta_values)**2
t0_correlation = np.sin(theta_values + 0.3)**2
redu_t_0_0correlation = t_0_0correlation[::3]  # Corrigido nome
```

```python
# Configurações gerais
plt.rcParams["mathtext.fontset"] = "cm"
fig, ax = plt.subplots()
marker_size = 12

# ========== FUNÇÕES AUXILIARES ==========
def generate_ticks(start, stop, step, labels=None):
    """Gera ticks principais e menores automaticamente"""
    ticks = np.arange(start, stop + step/2, step)  # +step/2 para incluir o último valor
    if labels is None:
        labels = [f"${tick:.2f}$" for tick in ticks]
    minor_ticks = []
    
    # Adiciona ticks menores entre os principais
    minor_step = step / 2
    minor_ticks = np.arange(start + minor_step, stop, minor_step)
    
    return ticks, labels, minor_ticks

# ========== CONFIGURAÇÃO DO GRÁFICO ==========
# Plot das curvas (mantido igual)
t_0theta_values = theta_values[::3]
redu_t_0_0correlation_values = t_0_0correlation_values[::3]

redu_t_0_0correation = t0_correlation[::3]


ax.plot(t_0theta_values, redu_t_0_0correlation_values, label=r"$C_{i}, t_0=0$", linestyle='-', color="darkgreen", marker="p", markeredgecolor='darkgreen', markeredgewidth=1.5, markersize=marker_size)
ax.plot(theta_values, correlation_values, label=r"$C_{i}, t_0=0.3$", linestyle='-', color="darkblue", marker="^", markeredgecolor='mediumblue', markeredgewidth=1.5, markersize=marker_size)
ax.plot(t_0theta_values, redu_t_0_0correation, label=r"$\delta n^2, t_0=0$", linestyle='-', color="saddlebrown", marker="X", markeredgecolor='sienna', markeredgewidth=1.5, markersize=marker_size)
ax.plot(theta_values, correlation, label=r"$\delta n^2, t_0=0.3$", linestyle='-', color="darkred", marker=markers[3], markeredgecolor='brown', markeredgewidth=1.5, markersize=marker_size)
```

```python
# ========== CONFIGURAÇÃO DOS EIXOS ==========
# Eixo Y
y_ticks, y_labels, y_minor_ticks = generate_ticks(0, 0.8, 0.2, 
                                                 [r"$0$", r"$0.2$", r"$0.4$", 
                                                  r"$0.6$", r"$0.8$"])
ax.set_yticks(y_ticks)
ax.set_yticklabels(y_labels, fontsize=22)
ax.set_yticks(y_minor_ticks, minor=True)

# Eixo X
x_ticks, x_labels, x_minor_ticks = generate_ticks(0, 2, 0.5,
                                                 [r"$0$", r"$0.5$", r"$1.0$",
                                                  r"$1.5$", r"$2.0$"])
ax.set_xticks(x_ticks)
ax.set_xticklabels(x_labels, fontsize=22)
ax.set_xticks(x_minor_ticks, minor=True)

# ========== ESTILIZAÇÃO ==========
# Ticks
ax.tick_params(axis='both', which='major', width=1.5, direction='in', 
               length=10, top=True, right=True)
ax.tick_params(axis='both', which='minor', width=1.5, direction='in', 
               length=5, top=True, right=True)

# Bordas
for spine in ax.spines.values():
    spine.set_linewidth(2)

# Legendas e limites
ax.set_ylabel(r"$\delta n^2$"r"$,C_{i}$", fontsize=26)
ax.set_xlabel(r"$\theta/\pi$", fontsize=26)
ax.legend(prop={'size':23}, loc='best', bbox_to_anchor=(0.72, 0.7), frameon=False, ncol=1)
ax.set_xlim(0, 2.0)
ax.set_ylim(-0.01, 0.8)

# Layout final
fig.set_size_inches(10, 7)
fig.tight_layout()
```
